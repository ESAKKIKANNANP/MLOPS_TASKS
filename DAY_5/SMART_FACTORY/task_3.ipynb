# Task 3: Save Model

**The Goal**: Bridge the gap between a local script and a deployable asset.

**The Action**: Take the best-performing multivariable model from the morning session and serialize it to the hard drive using `joblib`.

**The Research**: Write a short script in a separate file to load this saved model and predict the temperature for a mock "new" machine reading. This proves the model is detached from the training environment.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib
import os

# 1. Train the best-performing multivariable model
df = pd.read_csv('ai4i2020.csv')
features = ['Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]', 'Air temperature [K]']
X = df[features]
y = df['Process temperature [K]']

# Train on the entire dataset for the final deployable model
model = LinearRegression()
model.fit(X, y)

# 2. Serialize and save the model to the hard drive
model_filename = 'multivariable_model.joblib'
joblib.dump(model, model_filename)

print(f"Model successfully saved to '{model_filename}'")
print(f"File size: {os.path.getsize(model_filename)} bytes")
```
