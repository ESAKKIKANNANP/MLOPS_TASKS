```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
df=pd.read_csv('/content/canada_per_capita_income.csv')
year=df[['year']]
pci=df[['per capita income (US$)']]
model=linear_model.LinearRegression()
model.fit(year,pci)
model.predict([[2020]])
```
